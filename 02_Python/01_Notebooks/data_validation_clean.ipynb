# Customer Churn & Retention Analysis
## Data Validation

This notebook validates the raw workbook before cleaning and downstream analysis.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parents[1]
FILE_PATH = PROJECT_ROOT / '01_Data' / '01_Raw' / 'Customer_Churn_Retention_Raw_Data.xlsx'
print(FILE_PATH)
```

## 1. Load Raw Data

```python
excel_file = pd.ExcelFile(FILE_PATH)
print(excel_file.sheet_names)

customers = pd.read_excel(excel_file, sheet_name='customers')
subscriptions = pd.read_excel(excel_file, sheet_name='subscriptions')
payments = pd.read_excel(excel_file, sheet_name='payments')
customer_services = pd.read_excel(excel_file, sheet_name='customer_services')
support_tickets = pd.read_excel(excel_file, sheet_name='support_tickets')
data_dictionary = pd.read_excel(excel_file, sheet_name='Data_Dictionary')
```

## 2. Dataset Dimensions and Structure

```python
for name, df in {
    'Customers': customers,
    'Subscriptions': subscriptions,
    'Payments': payments,
    'Customer Services': customer_services,
    'Support Tickets': support_tickets,
    'Data Dictionary': data_dictionary,
}.items():
    print(f'{name}: {df.shape}')
```

## 3. Missing Value Analysis

```python
def missing_value_summary(df):
    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df) * 100).round(2)
    return pd.DataFrame({
        'Missing_Count': missing_count,
        'Missing_Percentage': missing_percentage
    }).sort_values('Missing_Count', ascending=False)

for name, df in {
    'Customers': customers,
    'Subscriptions': subscriptions,
    'Payments': payments,
    'Customer Services': customer_services,
    'Support Tickets': support_tickets,
    'Data Dictionary': data_dictionary,
}.items():
    print(f'\n--- {name} ---')
    display(missing_value_summary(df))
```

## 4. Duplicate Record Analysis

```python
for name, df in {
    'Customers': customers,
    'Subscriptions': subscriptions,
    'Payments': payments,
    'Customer Services': customer_services,
    'Support Tickets': support_tickets,
    'Data Dictionary': data_dictionary,
}.items():
    print(f'{name}: {df.duplicated().sum()} duplicate rows')
```
